This script aims to divide file baustellen_vorschau based on the geometry_type

```python

import pandas as pd
import os
```

```python
# Define input, output paths
current_dir = globals().get('current_dir', os.getcwd())
input_path = os.path.join(current_dir, '../data/mobidata/raw/csv_files/')
map_path = os.path.join(current_dir, '../data/map/Stadtteile_SHP/Stadtviertel_Karlsruhe.shp')
output_path = os.path.join(current_dir, '../data/mobidata/processed/')

# Define file path
input_file = 'baustellen_vorschau.csv'
input_file_path = os.path.join(input_path, input_file)
```

### 1. Load data

```python
# 1) Load data 
df = pd.read_csv(input_file_path)
```

```python
df
```

```python
geometry_types = df['geometry_type'].unique()
```

```python
geometry_types
```

```python
for geometry_type in geometry_types:
    df_filtered = df[df['geometry_type'] == geometry_type]
    
    # Create a variable name based on geometry type (e.g., df_Point, df_LineString)
    var_name = f'df_{geometry_type}'
    globals()[var_name] = df_filtered
    
    # save the filtered dataframe to a CSV file
    output_file = f'baustellen_vorschau_{geometry_type}.csv'
    output_file_path = os.path.join(input_path, output_file)
    df_filtered.to_csv(output_file_path, index=False)
```
